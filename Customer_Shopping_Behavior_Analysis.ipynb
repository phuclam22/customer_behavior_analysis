```python
import pandas as pd

df = pd.read_csv('customer_shopping_behavior.csv')
```

```python
df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

```python
df.isnull().sum()
```

```python
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ', '_')
df = df.rename(columns={'purchase_amount_(usd)' : 'purchase_amount'})
```

```python
df.columns
```

```python
labels = ['Young Adult', 'Adult', 'Middle-aged', 'Senior']
df['age_group'] = pd.qcut(df['age'], q=4, labels = labels)
```

```python
df[['age','age_group']].head(10)
```

```python
df['age_group'].value_counts()
```

```python
frequency_mapping = {
    'Fortnightly': 14,
    'Weekly': 7,
    'Monthly': 30,
    'Quarterly': 90,
    'Bi-Weekly': 14,
    'Annually': 365,
    'Every 3 Months': 90
}

df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
```

```python
df[['purchase_frequency_days','frequency_of_purchases']].head(5)
```

```python
from sqlalchemy import create_engine
from urllib.parse import quote_plus
import pandas as pd

# 1. Cấu hình kết nối bằng Windows Authentication (Không cần mật khẩu)
host = "localhost" 
database = "customer_behavior"
driver = quote_plus("ODBC Driver 17 for SQL Server")

# Chuỗi kết nối đã thêm Trusted_Connection=yes
engine = create_engine(f"mssql+pyodbc://{host}/{database}?driver={driver}&Trusted_Connection=yes")

table_name = "customer"

# 2. Vận chuyển dữ liệu
print("Đang vận chuyển dữ liệu vào SQL Server...")
df.to_sql(table_name, engine, if_exists="replace", index=False)
print(f"Thành công! Đã đổ dữ liệu vào bảng '{table_name}'.")

# 3. Kiểm tra xem SQL đã nhận được hàng chưa
print("\n--- Bản xem trước 5 dòng từ SQL Server ---")
check_df = pd.read_sql("SELECT TOP 5 * FROM customer;", engine)
check_df
```
